```python

import os
import pandas as pd

REQUIRED_FILES = [
    "courses.csv",
    "assessments.csv",
    "vle.csv",
    "studentInfo.csv",
    "studentRegistration.csv",
    "studentAssessment.csv",
    "studentVle.csv"
]

INPUT_DIR = "/kaggle/input"
DATA_DIR = None

for root, dirs, files in os.walk(INPUT_DIR):
    if all(name in files for name in REQUIRED_FILES):
        DATA_DIR = root
        break

if DATA_DIR is None:
    raise FileNotFoundError(
        "Could not find all 7 OULAD CSV files. "
        "Please attach the OULAD dataset in Kaggle."
    )

print("Dataset folder:", DATA_DIR)

for name in REQUIRED_FILES:
    path = os.path.join(DATA_DIR, name)
    print(f"{name}: {os.path.getsize(path) / 1_000_000:.2f} MB")
```

```python

import pandas as pd
import os

KEY = ["code_module", "code_presentation", "id_student"]

courses = pd.read_csv(
    os.path.join(DATA_DIR, "courses.csv"), na_values="?"
)

assessments = pd.read_csv(
    os.path.join(DATA_DIR, "assessments.csv"), na_values="?"
)

vle = pd.read_csv(
    os.path.join(DATA_DIR, "vle.csv"), na_values="?"
)

student_info = pd.read_csv(
    os.path.join(DATA_DIR, "studentInfo.csv"), na_values="?"
)

registration = pd.read_csv(
    os.path.join(DATA_DIR, "studentRegistration.csv"), na_values="?"
)

student_assessment = pd.read_csv(
    os.path.join(DATA_DIR, "studentAssessment.csv"), na_values="?"
)

student_vle = pd.read_csv(
    os.path.join(DATA_DIR, "studentVle.csv"),
    na_values="?",
    dtype={
        "id_student": "Int32",
        "id_site": "Int32",
        "date": "Int16",
        "sum_click": "Int16"
    }
)

tables = {
    "courses": courses,
    "assessments": assessments,
    "vle": vle,
    "student_info": student_info,
    "registration": registration,
    "student_assessment": student_assessment,
    "student_vle": student_vle
}

for name, df in tables.items():
    print(f"{name}: {df.shape[0]:,} rows, {df.shape[1]} columns")

print("\nAll 7 files loaded successfully!")
assert len(student_vle) == 10_655_280, "studentVle is incomplete, upload the full file again"
```

```python
# Check missing values in all tables

for name, df in tables.items():
    print(f"\n{name}")
    print(df.isnull().sum())
```

```python
# Check unique students

print("student_info:", student_info["id_student"].nunique())
print("registration:", registration["id_student"].nunique())
print("student_assessment:", student_assessment["id_student"].nunique())
print("student_vle:", student_vle["id_student"].nunique())

# Check unique modules

print("\nstudent_info:", student_info["code_module"].nunique())
print("courses:", courses["code_module"].nunique())
print("assessments:", assessments["code_module"].nunique())
print("vle:", vle["code_module"].nunique())

# Check unique presentations

print("\nstudent_info:", student_info["code_presentation"].nunique())
print("registration:", registration["code_presentation"].nunique())
print("courses:", courses["code_presentation"].nunique())
print("assessments:", assessments["code_presentation"].nunique())
print("vle:", vle["code_presentation"].nunique())
```

```python
# Check whether the student-course-presentation combination is unique

student_info_key = (
    student_info
    .groupby(["id_student", "code_module", "code_presentation"])
    .size()
)

print("Duplicate keys in student_info:",
      (student_info_key > 1).sum())

registration_key = (
    registration
    .groupby(["id_student", "code_module", "code_presentation"])
    .size()
)

print("Duplicate keys in registration:",
      (registration_key > 1).sum())
```

```python
# Create composite keys for student_info and registration

student_info_keys = set(
    zip(
        student_info["id_student"],
        student_info["code_module"],
        student_info["code_presentation"]
    )
)

registration_keys = set(
    zip(
        registration["id_student"],
        registration["code_module"],
        registration["code_presentation"]
    )
)

# Check unmatched keys

print("Keys in student_info but not registration:",
      len(student_info_keys - registration_keys))

print("Keys in registration but not student_info:",
      len(registration_keys - student_info_keys))
```

```python
# Check assessment join key

assessment_ids = set(assessments["id_assessment"])

student_assessment_ids = set(
    student_assessment["id_assessment"]
)

print(
    "Assessment IDs in student_assessment but not assessments:",
    len(student_assessment_ids - assessment_ids)
)

print(
    "Assessment IDs in assessments but not student_assessment:",
    len(assessment_ids - student_assessment_ids)
)
```

```python
# Check VLE site IDs

vle_ids = set(vle["id_site"])

student_vle_ids = set(student_vle["id_site"])

print(
    "VLE IDs in student_vle but not vle:",
    len(student_vle_ids - vle_ids)
)

print(
    "VLE IDs in vle but not student_vle:",
    len(vle_ids - student_vle_ids)
)
```

```python
# Check student IDs against the main student table

student_ids = set(student_info["id_student"])

assessment_student_ids = set(student_assessment["id_student"])
vle_student_ids = set(student_vle["id_student"])

print(
    "Students in student_assessment but not student_info:",
    len(assessment_student_ids - student_ids)
)

print(
    "Students in student_vle but not student_info:",
    len(vle_student_ids - student_ids)
)
```

```python
# Check whether student_vle duplicates are exact duplicates
print("Total rows:", len(student_vle))
print("Duplicate rows:", student_vle.duplicated().sum())

# Show some duplicate rows
print(student_vle[student_vle.duplicated(keep=False)].head(10))
```

```python
# Check the number of rows after removing exact duplicates

student_vle_clean = student_vle.drop_duplicates()

print("Original rows:", len(student_vle))
print("Rows after removing duplicates:", len(student_vle_clean))
print("Rows removed:", len(student_vle) - len(student_vle_clean))
```
